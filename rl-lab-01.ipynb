**Task 2: Create and Initialize an RL Environment**

```python

env = gym.make("CartPole-v1")


observation, info = env.reset(seed=42)

print("Initial Observation:", observation)
print("Info:", info)
```

**Task 3: Explore Observation and Action Spaces**

```python
print("Observation Space:", env.observation_space)
print("Type of Observation Space:", type(env.observation_space))

print("Action Space:", env.action_space)
print("Number of Possible Actions:", env.action_space.n)
```

**Task 4: Execute a Random Agent**

```python
observation, info = env.reset(seed=42)

total_reward = 0
step_count = 0
terminated = False
truncated = False

while not (terminated or truncated):
    action = env.action_space.sample()  # random action
    observation, reward, terminated, truncated, info = env.step(action)

    total_reward += reward
    step_count += 1

    print(f"Step: {step_count} | Action: {action} | Observation: {observation} | "
          f"Reward: {reward} | Terminated: {terminated} | Truncated: {truncated}")

print("\n--- Episode Summary ---")
print("Total Steps:", step_count)
print("Cumulative Reward:", total_reward)

env.close()
```
